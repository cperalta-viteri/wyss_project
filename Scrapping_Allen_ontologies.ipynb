```python
from allensdk.api.queries.ontologies_api import OntologiesApi
```

```python
import pandas as pd
import numpy as np
```

```python
asdk = OntologiesApi()
```

```python
asdk.get_atlases_table(atlas_ids=1)
```

```python
structures_adult_mouse = asdk.get_structures(structure_graph_ids=1)
structures_develop_mouse = asdk.get_structures(structure_graph_ids=17)
structures_adult_human = asdk.get_structures(structure_graph_ids=10)
structures_develop_human = asdk.get_structures(structure_graph_ids=16)
```

```python
from operator import itemgetter
```

```python
acronym_adult_mouse = list(map(itemgetter("acronym"), structures_adult_mouse))
acronym_develop_mouse = list(map(itemgetter("acronym"), structures_develop_mouse))
acronym_adult_human = list(map(itemgetter("acronym"), structures_adult_human))
acronym_develop_human = list(map(itemgetter("acronym"), structures_develop_human))
```

```python
acronym_df = pd.DataFrame(list(zip(acronym_adult_mouse, acronym_develop_mouse, acronym_adult_human, acronym_develop_human)),
             columns=["mouse_adult", "mouse_developing", "human_adult", "human_developing"])
```

```python
acronym_df.to_csv("notebooks/querying/acronym_df.csv", index=False)
```
